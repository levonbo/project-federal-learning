```python
import medmnist
from medmnist import INFO
import numpy as np
import matplotlib.pyplot as plt  

print("DOne")
```

```python
data_flag = 'pathmnist'  # change to bloodmnist, dermamnist, etc.
info = INFO[data_flag]
DataClass = getattr(medmnist, info['python_class'])

train_dataset = DataClass(split='train', size=128, download=True)
val_dataset = DataClass(split='val', size=64, download=True)
test_dataset = DataClass(split='test', size=64, download=True)

labels_train = train_dataset.labels
labels_val = val_dataset.labels
labels_test = test_dataset.labels

label_dict = info['label']
n_classes = len(label_dict)
is_multilabel = info['task'] == 'multi-label, binary-class'  # true for chestmnist
```

```python
fig, axes = plt.subplots(1, n_classes, figsize=(2 * n_classes, 2.5))

for class_idx in range(n_classes):
    if is_multilabel:
        # first sample where this class is present
        idx = np.where(labels_train[:, class_idx] == 1)[0][0]
    else:
        # first sample matching this class
        idx = np.where(labels_train[:, 0] == class_idx)[0][0]

    img, _ = train_dataset[idx]
    ax = axes[class_idx]
    ax.imshow(img, cmap='gray' if img.mode == 'L' else None)
    ax.set_title(label_dict[str(class_idx)], fontsize=8)
    ax.axis('off')

plt.tight_layout()
plt.savefig('images_medmnist/pathmnist_image_train.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, n_classes, figsize=(2 * n_classes, 2.5))

for class_idx in range(n_classes):
    if is_multilabel:
        # first sample where this class is present
        idx = np.where(labels_val[:, class_idx] == 1)[0][0]
    else:
        # first sample matching this class
        idx = np.where(labels_val[:, 0] == class_idx)[0][0]

    img, _ = val_dataset[idx]
    ax = axes[class_idx]
    ax.imshow(img, cmap='gray' if img.mode == 'L' else None)
    ax.set_title(label_dict[str(class_idx)], fontsize=8)
    ax.axis('off')

plt.tight_layout()
plt.savefig('images_medmnist/pathmnist_image_val.png', dpi=150, bbox_inches='tight')
plt.show()
```
